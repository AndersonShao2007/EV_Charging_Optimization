```python
import pandas as pd
import json

# Load the CSV file -- this is the raw data before applying the filter
folder = "/Users/fanyang/Documents/Family/kids/ScienceFair2024/calculation/Geology/"
file_name = "alt_fuel_stations_ev_charging_units_Jan_1_2025.csv" 
main_data = pd.read_csv(folder + file_name, low_memory=False)

# Load the GeoJSON file -- this is the json file that contains the ZIP codes
with open(folder + "Bay Area ZIP Codes_20250101.geojson", "r") as f:
    geojson_data = json.load(f)

# Extract ZIP codes from the GeoJSON file
# Assuming the ZIP codes are stored in a property called "ZIP" or similar
unique_zips = set()
for feature in geojson_data["features"]:
    zip_code = feature["properties"].get("zip")  # Replace "ZIP" with the actual property name
    if zip_code:
        unique_zips.add(zip_code)
        
print(unique_zips)

```

```python
# Ensure the "ZIP" column exists in the main dataset
if "ZIP" not in main_data.columns:
    raise ValueError("The main dataset does not contain a 'ZIP' column.")

# Filter the main dataset to include only rows with ZIP codes in the unique ZIP list
filtered_data = main_data[main_data["ZIP"].isin(unique_zips)]

# Save the filtered data to a new CSV file (optional)
filtered_data.to_csv(folder + "filtered_data.csv", index=False)

# Print the filtered data
print("Filtered Data:")
print(filtered_data)
```
