calculate the actual centroid coordinates for each ZIP zone in the GeoJSON file

```python
import geopandas as gpd

# Load the GeoJSON file
folder = "/Users/fanyang/Documents/Family/kids/ScienceFair2024/calculation/geology/"
geo_data = "Bay Area ZIP Codes_20250101.geojson"
gdf = gpd.read_file(folder+geo_data)

# Check the current CRS
print("Original CRS:", gdf.crs)

# Reproject the geometries to a projected CRS (e.g., UTM Zone 10N for the Bay Area)
gdf = gdf.to_crs(epsg=32610)  # EPSG:32610 is UTM Zone 10N

# Calculate centroids in the projected CRS
gdf['centroid'] = gdf.geometry.centroid

# Reproject the centroids back to the geographic CRS (WGS84) for latitude/longitude
gdf['centroid'] = gdf['centroid'].to_crs(epsg=4326)  # EPSG:4326 is WGS84

# Extract latitude and longitude from the centroid
gdf['Latitude'] = gdf.centroid.y
gdf['Longitude'] = gdf.centroid.x

# Drop the 'centroid' column if you don't need it anymore
gdf = gdf.drop(columns=['centroid'])

# Check the updated GeoDataFrame
print(gdf.head())

# Save the updated GeoDataFrame to a new file (optional)
gdf.to_file(folder+"Bay Area ZIP Codes_with_centroids.geojson", driver='GeoJSON')
```
